```python
df["revenue"] = df["quantity"]*df["price"]
df.head()
```

```python
df.describe()
#This helps to understand: Average price, Average revenue sold, Revenue distribution
```

```python
df["order_date"]=pd.to_datetime(df["order_date"])
df["month"]=df["order_date"].dt.month
df["year"]=df["order_date"].dt.year
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
```

```python
category_revenue=df.groupby("category_name")["revenue"].sum()
category_revenue.plot(kind="bar")
plt.title("revenue by category")
plt.ylabel("revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["price"], df["quantity"]) 
plt.xlabel("price")
plt.ylabel("quantity sold")
plt.title("price vs quantity sold")
plt.show()
#This will show wheather higher prices lead to lower sales.
```

```python
df[["price","quantity"]].corr()
```

```python
plt.scatter(df["price"],df["revenue"])
plt.xlabel("price")
plt.ylabel("revenue")
plt.title("price vs revenue")
plt.show()
#This helps identify wheather increasing price increases revenue
```

```python
top_products=df.groupby("product_name")["revenue"].sum().sort_values(ascending=False)
print(top_products.head(10))
```

```python
avg_rating=df.groupby("category_name")["review_score"].mean()
print(avg_rating.sort_values(ascending=False))
#This gives business insights.
```

Insights:
- Electronic generated the highest revenue.
- Higher prices did not always reduce sales.
- Some products generated high revenue despite higher prices.
- Certain categories received better customer ratings.

```python
payment_revenue=df.groupby("payment_method")["revenue"].sum()
payment_revenue.plot(kind="bar")
plt.title("revenue by payment method")
plt.ylabel("revenue")
plt.show()
```

```python
df["payment_method"].value_counts().head()
```

```python
gender_revenue=df.groupby("gender")["revenue"].sum()
print(gender_revenue)
```

```python
city_revenue=df.groupby("city")["revenue"].sum().sort_values(ascending=False)
print(city_revenue.head())    
```

```python
import seaborn as sns
sns.heatmap(df[["price","quantity","revenue","age","review_score"]].corr(),annot=True)
plt.title("correlation matrix")
plt.show()
```

Key Findings:
1. Highest Revenue Generating Category= Sports and Outdoors
2. Top Selling Product= Smartphones
3. Most Preferred Payment Method= Cash on Delievry
4. Relationship between price and quantity = A very weak relationship
5. Top Revenue Generating City= Port Melissaborough

Business Recommendation:
1. Focus marketing efforts on the Sports and Outdoors category.
2. Promote Smartphones to increase sales further.
3. Continue offering Cash on Delivery as a preferred payment option.
4. Review pricing strategies to improve revenue.
5. Target customers in Port Melissaborough with additional marketing campaigns.
